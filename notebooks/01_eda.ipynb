```
import json, pandas as pd
from pathlib import Path
ROOT = Path('..')
report = json.loads((ROOT/'reports/dataset_report.json').read_text())
{k: v for k, v in report.items() if k not in ('class_counts', 'corrupted_files')}
```

```
manifest = pd.read_csv(ROOT/'data/processed/manifest.csv')
manifest.groupby('split').size()
```

```
counts = manifest.groupby(['label_name', 'split']).size().unstack(fill_value=0)
counts.assign(total=counts.sum(axis=1)).sort_values('total', ascending=False)
```

```
manifest[['width', 'height']].describe()
```

```
# classes with few samples are where macro-F1 will be noisy
counts.sum(axis=1).sort_values().head(15)
```

```
import matplotlib.pyplot as plt
from PIL import Image
root = Path(json.loads((ROOT/'data/processed/split_info.json').read_text())['source_root'])
sample = manifest.sample(12, random_state=0)
fig, axes = plt.subplots(3, 4, figsize=(10, 8))
for ax, (_, r) in zip(axes.flat, sample.iterrows()):
    ax.imshow(Image.open(root/r.path).convert('RGB')); ax.set_title(r.label_name, fontsize=7); ax.axis('off')
```
